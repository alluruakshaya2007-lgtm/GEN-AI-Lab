```python

from collections import Counter

words = {
    "LOW": 5,
    "LOWER": 2,
    "NEWEST": 3,
    "WILDEST": 6
}

# Step 1: Initial tokenization
tokens = {
    word: list(word.lower())
    for word in words
}

print("=" * 50)
print("BPE TOKENIZATION")
print("=" * 50)

print("\nINITIAL TOKENS:")
for word, token_list in tokens.items():
    print(f"{word} (Frequency {words[word]}): {token_list}")


# Function to calculate adjacent pair frequencies
def calculate_pairs(tokens):
    pair_freq = Counter()

    for word, token_list in tokens.items():
        for i in range(len(token_list) - 1):
            pair = (token_list[i], token_list[i + 1])
            pair_freq[pair] += words[word]

    return pair_freq


# Step 2: Repeatedly merge highest-frequency pairs
iteration = 1

while True:

    pair_freq = calculate_pairs(tokens)

    if not pair_freq:
        break

    print("\n" + "=" * 50)
    print(f"ITERATION {iteration}")
    print("=" * 50)

    print("\nADJACENT PAIR FREQUENCIES:")

    for pair, freq in pair_freq.items():
        print(f"{pair} = {freq}")

    # Find highest frequency pair
    highest = max(pair_freq.values())

    print("\nHighest Frequency:", highest)

    # Handle ties by selecting the first pair
    best_pair = next(
        pair for pair, freq in pair_freq.items()
        if freq == highest
    )

    print("Selected Pair:", best_pair)
    print("Merge:", best_pair[0], "+", best_pair[1],
          "=", best_pair[0] + best_pair[1])

    # Step 3: Update tokens
    for word in tokens:
        new_tokens = []
        i = 0

        while i < len(tokens[word]):

            if (i < len(tokens[word]) - 1 and
                (tokens[word][i], tokens[word][i + 1]) == best_pair):

                new_tokens.append(best_pair[0] + best_pair[1])
                i += 2

            else:
                new_tokens.append(tokens[word][i])
                i += 1

        tokens[word] = new_tokens

    print("\nUPDATED TOKENS:")

    for word, token_list in tokens.items():
        print(f"{word}: {token_list}")

    iteration += 1

print("\n" + "=" * 50)
print("FINAL TOKENS")
print("=" * 50)

for word, token_list in tokens.items():
    print(f"{word}: {token_list}")
```

```python

from collections import Counter

words = {
    "LOW": 5,
    "LOWER": 2,
    "NEWEST": 3,
    "WILDEST": 6
}

char_freq = Counter()
pair_freq = Counter()

for word, freq in words.items():
    chars = list(word.lower())

    for ch in chars:
        char_freq[ch] += freq

    for i in range(len(chars) - 1):
        pair = (chars[i], chars[i + 1])
        pair_freq[pair] += freq

print("WORDPIECE - Pair Scores")
print("------------------------")

scores = {}

for pair, freq in pair_freq.items():
    a, b = pair

    score = freq / (char_freq[a] * char_freq[b])
    scores[pair] = score

    print(f"{pair}: {freq}/({char_freq[a]}*{char_freq[b]}) = {score:.4f}")

highest = max(scores.values())

print("\nHighest WordPiece Score:", round(highest, 4))

print("Highest Scoring Pair(s):")

for pair, score in scores.items():
    if score == highest:
        print(pair, "=", round(score, 4))
```
